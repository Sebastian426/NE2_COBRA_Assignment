```python
import numpy as np
import matplotlib.pyplot as plt
```

# Linear Power Rating of 1 Rod

```python
def one_rod_linear_power(z, qmax, L):
    return qmax* np.sin((np.pi * z)/L)
```

```python
dh = 3.66/18
Q = 3320e6/(193*264)
Length = 3.66
qmax = Q* np.pi/(2*Length)


z_coordinates = []

for i in range(18):
    z_i = (i+0.5)*dh
    z_coordinates.append(z_i)

z_coordinates = np.array(z_coordinates)

power_slices = one_rod_linear_power(z_coordinates, qmax, Length)

print(power_slices)
```

```python
plt.figure(1)

plt.scatter(z_coordinates, power_slices, color = "black", marker = 'D', s = 3)

plt.xlabel("Z Coordinates")
plt.ylabel("Power (W)")
plt.axhline(qmax, color = 'red', linestyle = 'dashed')

plt.show()
```
